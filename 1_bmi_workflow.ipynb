```python
from langgraph.graph import StateGraph,START, END
from typing import TypedDict
```

```python
# Define State 
class BMIState(TypedDict):
    Weight_kg : float
    height_m : float
    bmi : float

```

```python
def calculate_bmi(state: BMIState) -> BMIState:  #type hunting for the function
    weight = state['Weight_kg']
    height = state['height_m']
    bmi = weight / (height ** 2)
    state['bmi'] = bmi
    return state
```

```python
# Define your state graph

graph = StateGraph(BMIState)


#add nodes to the graph
graph.add_node('calculate_bmi', calculate_bmi)

# add edges to the graph
graph.add_edge(START, 'calculate_bmi')
graph.add_edge('calculate_bmi', END)

# compile the graph 
workflow = graph.compile()
```

```python
# Execute the graph
initial_state = {'Weight_kg': 70, 'height_m': 1.75} # imp

final_state = workflow.invoke(initial_state)

print(final_state)  # This will print the final state with the calculated BMI
```
